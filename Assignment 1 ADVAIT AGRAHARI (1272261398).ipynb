```python
#list Data Structure
my_list = ["Apple","Banana","Mango",]
print("Initial List :" , my_list)

#Append
my_list.append("Watermelon")
print("List after Appending :", my_list)

#Remove
my_list.remove("Banana")
print("List after Removal:" , my_list)

#Dictionary 
dicti = {"Name " : "Advait" , "age": 19 }
print("Initial Dictionary:" , dicti)

#Append
dicti ["City"] = "PUNE"
print("Updated Dictionary :", dicti)

#Remove
del dicti["age"]
print("Updated Dictionary:", dicti)

#Tuple
tup = (10,20,30,)
print("Initial Tuple:",tup)

#Append
temp_tup = list(tup)
temp_tup.append(40)
tup = tuple(temp_tup)
print("Appended Tuple :" ,temp_tup)

#Remove
temp_tup = list(tup)
temp_tup.remove(20)
tup = tuple(temp_tup)
print("Updated Tuple:",tup)
```
